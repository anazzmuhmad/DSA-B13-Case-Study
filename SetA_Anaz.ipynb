```python
#Question 3

class Product:
    def __init__(self, product_id, product_name, price, stock):
        self.product_id = product_id
        self.product_name = product_name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self):
        self.cart = {}

    def add_product(self, products, product_id, quantity):
        if product_id not in products:
            print("Product not found")
        elif quantity <= 0:
            print("Invalid quantity")
        elif quantity > products[product_id].stock:
            print("Insufficient stock")
        else:
            if product_id in self.cart:
                self.cart[product_id] += quantity
            else:
                self.cart[product_id] = quantity

            products[product_id].stock -= quantity

    def remove_product(self, products, product_id, quantity):
        if product_id not in self.cart:
            print("Product not in cart")
        elif quantity > self.cart[product_id]:
            print("Quantity not available in cart")
        else:
            self.cart[product_id] -= quantity
            products[product_id].stock += quantity

            if self.cart[product_id] == 0:
                del self.cart[product_id]

    def calculate_total(self, products):
        total = 0

        for product_id in self.cart:
            quantity = self.cart[product_id]
            price = products[product_id].price
            total = total + (price * quantity)

        return total

    def apply_discount(self, total):
        if total < 5000:
            discount = 0
        elif total < 10000:
            discount = 5
        elif total < 20000:
            discount = 10
        else:
            discount = 15

        discount_amount = total * discount / 100
        final_amount = total - discount_amount

        return discount, discount_amount, final_amount

    def display_cart(self, products):
        if len(self.cart) == 0:
            print("Cart is empty")
            return

        for product_id in self.cart:
            quantity = self.cart[product_id]
            product = products[product_id]

            subtotal = product.price * quantity

            print(
                product.product_name,
                "- Quantity:", quantity,
                "- Subtotal:", subtotal
            )

        total = self.calculate_total(products)
        discount, discount_amount, final_amount = self.apply_discount(total)

        print()
        print("Total Amount:", total)
        print("Discount:", discount, "%")
        print("Discount Amount:", discount_amount)
        print("Final Amount:", final_amount)


products = {}

n = int(input())

for i in range(n):
    data = input().split()

    product_id = int(data[0])
    product_name = data[1]
    price = float(data[2])
    stock = int(data[3])

    products[product_id] = Product(
        product_id, product_name, price, stock
    )


cart = ShoppingCart()

operations = int(input())

for i in range(operations):
    data = input().split()

    if data[0] == "ADD":
        product_id = int(data[1])
        quantity = int(data[2])

        cart.add_product(products, product_id, quantity)

    elif data[0] == "REMOVE":
        product_id = int(data[1])
        quantity = int(data[2])

        cart.remove_product(products, product_id, quantity)

    elif data[0] == "TOTAL":
        total = cart.calculate_total(products)
        print("Total:", total)

    elif data[0] == "DISPLAY":
        cart.display_cart(products)
```
