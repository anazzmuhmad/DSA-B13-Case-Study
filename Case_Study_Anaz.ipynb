```python
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}

]

total_employees = len(employees)
print("Total employees : ", total_employees)

total_salary = sum(employee["salary"] for employee in employees)

average_salary = total_salary / total_employees
print("\nAverage salary : ", average_salary)

highest_salary = max(employee["salary"] for employee in employees)
print("Highest salary : ", highest_salary)

lowest_salary = min(employee["salary"] for employee in employees)
print("Lowest salary : ", lowest_salary)

print("\nEmployees with salary > 65000 and Rating >= 4.5")


for employee in employees:
   if employee["salary"] > 65000 and employee["rating"] >= 4.5:
       print(employee["name"])

employee_dictionary = {}

for employee in employees:
    employee_dictionary[employee["id"]] = employee["name"]
print("\nEmployee Dictionary")
print(employee_dictionary)

names = []

for employee in employees:
    names.append(employee["name"])

names.sort()
names.reverse()

print("\nFirst Three Employee Names")

for name in names[:3]:
    print(name)

employee_ids = ()

for employee in employees:
    employee_ids = employee_ids + (employee["id"],)

print("\nEmployee ID Tuple")
print(employee_ids)
print("\nMaximum Employee ID:", max(employee_ids))
print("Minimum Employee ID:", min(employee_ids))

departments = set()

for employee in employees:
    departments.add(employee["department"])

print("\nDepartments")
print(departments)

search_id = int(input("\nEnter Employee ID : "))

found = False

for employee in employees:
    if employee["id"] == search_id:
        print("\nEmployee Found")
        print("ID : ", employee["id"])
        print("Name : ", employee["name"])
        print("Department : ", employee["department"])
        print("Salary : ", employee["salary"])
        print("Rating : ", employee["rating"])
        found = True

if found == False:
    print("Employee Not Found")


```
